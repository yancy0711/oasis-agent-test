# OASIS agent quickstart

Start the API: 

python -m uvicorn api.main:app --port 8000

Minimal bounded-agent request:

```python
import requests
payload = {"prompt": "Select 10 candidate cooling facilities in Phoenix within a 15-minute walk."}
response = requests.post("http://127.0.0.1:8000/v1/parse", json=payload, timeout=60)
response.raise_for_status()
response.json()
```

The response contains the bounded task, routing disposition, missing inputs, and any route-origin candidates. Use the web UI for place confirmation and full GIS output.
